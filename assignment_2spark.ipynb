```python
from pyspark.sql import SparkSession

# Initialize SparkSession running locally across available cores
spark = SparkSession.builder \
    .master("local[*]") \
    .appName("MEM6287B00_Part2") \
    .getOrCreate()

sc = spark.sparkContext
print(f"Master (Active Cluster): {sc.master}")
print(f"Spark Version: {spark.version}")
```

```python
# Load all .text files from the data directory into an RDD
corpus_rdd = sc.textFile("data/*.text")
print(f"Corpus RDD: {corpus_rdd}")
```

```python
# Count total lines across all files in the RDD
total_lines = corpus_rdd.count()
print(f"Total number of lines across all files: {total_lines}")
```

```python
import re

# Tokenize into individual words, strip punctuation, and convert to lowercase
words_rdd = corpus_rdd.flatMap(lambda line: re.findall(r"\b[a-zA-Z0-9']+\b", line.lower()))

# Filter for the target word and count
antibiotics_count = words_rdd.filter(lambda word: word == "antibiotics").count()
print(f"Total occurrences of 'antibiotics': {antibiotics_count}")
```
